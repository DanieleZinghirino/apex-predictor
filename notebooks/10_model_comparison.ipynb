# Apex Predictor, Confronto finale: XGBoost vs Deep Learning

Sintesi di 4 esperimenti di deep learning (MLP con embedding, LSTM 
sulla sequenza grezza, TabNet, GNN su relazioni compagno di squadra), 
tutti valutati con la stessa metodologia (split temporale 2025-2026, 
ricerca soglia ottimale su F1) usata per XGBoost.

## Risultati

| Modello | Precision | Recall | F1 |
|---|---|---|---|
| **XGBoost (campione)** | 0.636 | 0.820 | **0.717** |
| TabNet | 0.656 | 0.775 | 0.711 |
| LSTM ibrida | 0.655 | 0.703 | 0.678 |
| GNN (GraphSAGE) | 0.634 | 0.703 | 0.667 |
| MLP + embedding | 0.552 | 0.811 | 0.657 |

## Conclusione

XGBoost resta il modello campione. Nessuna delle 4 architetture di 
deep learning ha superato il baseline, con TabNet a distanza minima 
(-0.006 F1, sostanzialmente equivalente). Il risultato è coerente 
con la letteratura consolidata su deep learning per dati tabellari: 
con ~8500 righe di training (e solo ~400 "grafi" per l'esperimento 
GNN), il vantaggio classico del deep learning, la capacità di 
apprendere rappresentazioni complesse da grandi quantità di dati,
non ha modo di manifestarsi appieno; gradient boosting resta più 
efficiente e meno soggetto a overfitting su dataset di questa scala.

## Osservazioni qualitative che vale la pena portare avanti

- **TabNet** ha mostrato il ranking di importanza feature più coerente 
  con l'analisi SHAP di XGBoost (stesse 4 feature principali, ordine 
  quasi identico), buona conferma incrociata che il segnale dominante 
  del problema (griglia, forma recente, classifica) è reale e non un 
  artefatto di un singolo algoritmo.
- **LSTM** è l'unico modello che non richiede feature di forma 
  calcolate a mano (EWMA/mediana),potenzialmente più scalabile se il 
  dataset crescesse di molto in futuro.
- **GNN**, l'esperimento più penalizzato dalla scarsità di dati (411 
  grafi vs migliaia di righe per gli altri), suggerisce che l'idea di 
  modellare esplicitamente le relazioni compagno di squadra ha 
  potenziale ma richiederebbe un dataset più ampio (più stagioni 
  storiche, o un grafo con più tipi di archi) per essere sfruttata 
  davvero.

**Decisione**: XGBoost rimane il modello in produzione. Questo 
esercizio dimostra però un percorso di validazione rigoroso, quattro 
architetture diverse testate con lo stesso rigore metodologico, 
nessuna adottata "per moda" senza controprova sui dati.
